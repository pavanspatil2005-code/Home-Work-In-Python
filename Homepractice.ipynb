## 1) Reverse a String :-
- I reverse the given string and display it.

```python
a="Pavan"
reverse= a[::-1]
print("this is a string", a)
print("this is a reverse string", reverse)
```

## 2) Palindrome String :-
- first reverse the string then compare reverse and original string if same then palindrome if not then no palindrome

```python
string = "pavan"
if string == string[::-1]:
    print("the string is palindrome")
else:
    print("the string is not palindrome")

name="Level"

if name.lower() == name.lower()[::-1]:
    print("the name is palindrome")
else:
    print("the name is not palindrome")
```

## 3) Count Vowels in string :-


```python
s= "vowels in the counts"
count=0
for i in s:
    if i in "aeiouAEIOU":
        count+=1
print("Number of vowels:", count)
```

## 4) Count Consonants in string :-

```python
s= "My Name IS PaVan"
count=0
for i in s:
    if i.isalpha() and i not in "aeiouAEIOU":
        count+=1
print("Number of consonants:", count)
```

## 5) Count both vowels and consonants

```python
string= "Pavan Patil"
vowels=0
consonants=0
for i in string:
    if i.isalpha():
        if i.lower() in "aeiou":
            vowels+=1
        else:
            consonants+=1
print("Number of vowels:", vowels)
print("Number of consonants:", consonants)
```

## 6) Counts Digits in a String

```python
s = "My Phone Number is 1234567890"
count=0
for i in s:
    if i.isdigit():
        count+=1
print("Number of digits:", count)
```

## 7) Count Spaces in string 

```python
a= "Pavan     "
count=0
for i in a:
    if i.isspace():
        count+=1
print("spaces in characters:", count)
```
