* WAP to get a unique element list form above list without chengining data position

```python
l = [5,4,5,4,2,1,3,1,6,4,2,3,5,7,6]
unique_data=[]

for data in l:
    if data not in unique_data:
        unique_data.append(data)
print(unique_data)
```

* find the freqeuncy of every element from above input list in dictionary

```python
l = [5,4,5,4,2,1,3,1,6,4,2,3,5,7,6]

freq_list={}

for data in l:
    if data in freq_list:
        freq_list[data] == freq_list[data] + 1
    else:
        freq_list[data] = 1
print(f"how meny times repeated each element: {freq_list}")
```
